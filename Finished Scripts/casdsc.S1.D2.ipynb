### Merging Data

Merging data is an incredibly common thing you will need to do when working with data. Python handles this in a variety of ways depending on how you need to merge. It is important that you understanding if you need a many to one, many to many, inner join, outer join, etc.

This example is a many to one, as in merging observations that share a common variable.

```python
import pandas as pd
df1 = pd.DataFrame({'Name': ['a1','a2','a3','b1','b2','b3'], 'data1': range(6)})
df1
```

```python
df2 = pd.DataFrame({'Name': ['a1','b1','b2'], 'data2': range(3)})
df2
```

```python
df = pd.merge(df1,df2)
df
```

Another example

```python
df1 = pd.DataFrame({'id': ['var1','var2','var23','var4'], 'data1': range(4)})
df1
```

```python
df2 = pd.DataFrame({'id': ['var1','var3','var16'], 'data2': ['tree','hat',12]})
df2
```

```python
df = pd.merge(df1,df2)
df
```

Python defaults to using a commonly shared column to merge if there is one. Above the commonly shared columns were key. If there isn't, you will get an error.

```python
df3 = pd.DataFrame({'Id Survey 1': ['a1','a2','a3','b1','b2','b3','c1'], 'data1': range(7)})
df4 = pd.DataFrame({'Id Survey 2': ['a1','b2','c1'], 'data2': range(3)})
```

```python
df3
```

```python
df4
```

```python
pd.merge(df3,df4)
```

The way to deal with this is to specify which column you want to merge on. left is the first df specified, right is the second.

```python
pd.merge(df3,df4, left_on = 'Id Survey 1', right_on='Id Survey 2')
```

Another example.

```python
df5 = pd.DataFrame({'lkey': [16,15,65,'b1',12,'b3',99], 'data1': range(7)})
df6 = pd.DataFrame({'rkey': [16,'b1','c1'], 'data2': range(3)})
print(df5)
print(df6)
```

```python
pd.merge(df5,df6, left_on = 'lkey', right_on='rkey')
```

What we have been doing are inner joins. That means we lose non-shared data. We can specify as outer to include all variables.

```python
df3 = pd.DataFrame({'lkey': ['a1','a2','a3','b1','b2','b3','c1'], 'data1': range(7)})
df4 = pd.DataFrame({'rkey': ['a1','b2','c6'], 'data2': range(3)})
print(df3)
print(df4)
```

```python
df5 = pd.merge(df3,df4, left_on = 'lkey', right_on='rkey', how = 'outer')
df5
```

### More Merging Examples

```python
import pandas as pd
import numpy as np

left = pd.DataFrame({
    'id':[1,2,2,3,4],
    'year':[2022,2022,2023,2023,2023],
    'x':[10,20,25,30,40]
})
right = pd.DataFrame({
    'id':[1,2,2,4,5],
    'year':[2022,2022,2023,2023,2024],
    'y':[100,200,250,400,500]
})

left, right
```

Merge on multiple keys (id, year)

```python
merged_inner = pd.merge(left, right, on=['id','year'], how='inner')
merged_inner
```

```python
merged_left = pd.merge(left, right, on=['id','year'], how='left')
merged_left
```

```python
merged_right = pd.merge(left, right, on=['id','year'], how='right')
merged_right
```

```python
merged_outer = pd.merge(left, right, on=['id','year'], how='outer', indicator=True)
merged_outer
```

Multiple aggregations per group using a list to subset on DF on then creating new columns.

```python
agg_df = (sales.groupby(['store','dept']).agg(
        revenue_sum=('revenue','sum'),
        revenue_mean=('revenue','mean'),
        transactions_sum=('transactions','sum')
    ).reset_index() # we need this to reset the index to give us normal columns instead of nested ones
)
agg_df.head()
```

```python
agg_df.describe()
```

Filtering groups after aggregation and looking for ones that appear more than 5 times in the dataset

Run this command to make a fake dataset.

```python
import numpy as np
import pandas as pd

np.random.seed(1234)

n = 1200
df = pd.DataFrame({
    "id": np.arange(n),
    "group": np.random.choice([f"G{i}" for i in range(1, 21)], size=n),
    "region": np.random.choice(["North","South","West"], size=n),
    "sector": np.random.choice(["Public","Private","Nonprofit","Academic"], size=n),
    "date": pd.to_datetime("2024-01-01") + pd.to_timedelta(np.random.randint(0, 180, size=n), unit="D"),
})

df["x1"] = np.random.normal(0, 1, size=n)
df["x2"] = np.random.lognormal(mean=0.3, sigma=0.6, size=n)
df["x3"] = np.random.binomial(1, 0.45, size=n)
df["experience"] = np.random.randint(0, 21, size=n)
df["treatment"] = np.random.binomial(1, 0.4, size=n)

region_effect = {"North": 0.2, "South": -0.3, "West": 0.0}
sector_effect = {"Public": 0.4, "Private": 0.0, "Nonprofit": -0.2, "Academic": 0.1}

eps = np.random.normal(0, 1.0, size=n)
df["y_cont"] = (
    2.0
    + 1.2*df["x1"]
    - 0.7*df["x2"]
    + 0.6*df["x3"]
    + 0.03*df["experience"]
    + 0.5*df["treatment"]
    + df["x1"] * (df["treatment"]) * 0.4
    + df["region"].map(region_effect)
    + df["sector"].map(sector_effect)
    + eps
)

linpred = (
    -0.6
    + 0.6*df["x1"]
    - 0.4*df["x2"]
    + 0.5*df["x3"]
    + 0.02*df["experience"]
    + 0.7*df["treatment"]
    + df["region"].map({k:v*0.8 for k,v in region_effect.items()})
    + df["sector"].map({k:v*0.8 for k,v in sector_effect.items()})
)
p = 1 / (1 + np.exp(-linpred))
df["y_bin"] = np.random.binomial(1, p)

eta = np.random.normal(0, 0.8, size=n)
df["wage"] = (
    15
    + 2.0*df["x1"]
    + 0.8*df["experience"]
    - 0.5*df["x2"]
    + 1.0*df["treatment"]
    + df["region"].map(region_effect)
    + df["sector"].map(sector_effect)
    + eta
)

df.head()
```

### Text Examples

There is so much we can do with text. This is a mild taste of it. We are going to bring in four packages, pandas, matplotlib, and statsmodels we have used before, and re is a new one.

```python
import re
import statsmodels.formula.api as smf
import pandas as pd
import matplotlib.pyplot as plt
```

We are looking at transcripts for States of the Union from 1790 - 2018. They are already downloaded and in the folder. Data is from Brian Weinstein https://github.com/BrianWeinstein/state-of-the-union

```python
df = pd.read_csv('transcripts.csv', sep = ',') 
df.head()
```

Look at how often presidents appear in the dataset.

```python
df['president'].value_counts()
```

This lets us count how many times the word apple appears in each speech. str.count looks for the word in the text (in this case the column transcript) and counts each instance of it. It is case-sensitive so we should lower case our transcripts.

```python
df['lower'] = df['transcript'].str.lower()
```

```python
df.head()
```

```python
df['count'] = df['lower'].str.count('apple')
df.head()
```

```python
df['count'].value_counts()
```

This tells us that in 230 speeches the word apple is not used. It is used twice in 13 of them and once in one of them.

We can build a loop that does the same thing but for any number of words. This is a very basic code for dictionary based processing.

First we create an empty column, which will be where we put our counts for each individual word added together.

```python
americaWords = ['freedom','liberty','rights','america',"military"]

df['words'] = 0

for i in americaWords:
    df['words'] = df['lower'].str.count(i) + df['words']
df.head()
```

```python
df['words'].value_counts()
```

Let's plot what we see.

```python
plt.hist(df['words'])
plt.show()
```

```python
plt.hist(df['words'])
plt.xlabel('American Words')
plt.ylabel('Counts')
plt.title("American Words in SOTU Addresses")
plt.show()
```

We can now run some models to see which president uses those words the most. We put president inside of a C() to indicate a categorical variable.

```python
results = smf.ols('words ~ C(president)', data=df).fit()
print(results.summary())
```

While that is interesting, that may not be super useful because speeches vary in length. Let's check the proportion of America words in each speech. First we need to get the word counts for each.

```python
df['wc'] = df['lower'].str.split().str.len()
df.head()
```

Then divide America words by the total number of words.

```python
df['americaProp'] = df['words']/df['wc']
df.head()
```

```python
df['americaProp'].value_counts()
```

New results.

```python
results = smf.ols('americaProp ~ C(president)', data=df).fit()
print(results.summary())
```

```python
df.head()
```

That is pretty exciting! We can also think about year, has our proportion increased year over year? There are a few ways to do this. The first is easy because we know the list is in sequential order. We insert a range of 244 to 0 and use -1 as the 'step', or basically decrease the number by 1 each time.

```python
df['order'] = range(244, 0, -1)
df.tail()
```

Now we can use order as our independent variable on americaProp. We see that, overtime, there is an increase in the proportion of words that are part of our dictionary.

```python
results = smf.ols('americaProp ~ order', data=df).fit()
print(results.summary())
```

We can also plot this. We see that around speech 150, there is a huge increase in the usage of these words in speeches.

```python
plt.figure(figsize=(8, 6), layout='constrained')
plt.plot(df['order'], df['americaProp'], label='americaProp')
plt.xlabel('Speech Order')
plt.ylabel('Proportion')
plt.title("americaProp Over Time")
plt.legend()
plt.show()
plt.save("name of file.pdf")
```

What about if we don't know the order or if it is random across our data? Here we can use the date variable and pandas. First we set the "date" column to a datetime object. This is sort of like when we set strings, integers, or floats. Python and pandas will treat this as a date instead of just text.

```python
df['date'] = pd.to_datetime(df['date'])
df.head()
```

Next we use pandas to strip out the year from the 'date' column. We set this as a new column called 'year'.

```python
df['year'] = df['date'].dt.year
df.head()
```

Then we just run our regression using year. We will get slightly different results because multiple speeches can be made in a year.

```python
df['year'].value_counts()
```

```python
results = smf.ols('americaProp ~ year', data=df).fit()
print(results.summary())
```

### Network Analysis (Basics)
We will use small examples to understand nodes, edges, degrees, and centrality.

```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

edges = pd.DataFrame({
    'source': [
        'A','A','B','C','D','E','F','G','G','H','H',  
        'B','D','E','F','I','J','K','L','M','N','O',  
        'P','Q','R','S','T','U','V','W','X','Y','Z'   
    ],
    'target': [
        'B','C','C','D','E','F','C','H','I','I','J',  
        'G','H','I','J','K','L','M','N','O','P','Q',  
        'Q','R','S','T','U','V','W','X','Y','Z','A'  
    ]}
)
edges
```

Build a graph from an edge list

```python
G = nx.from_pandas_edgelist(edges, source='source', target='target', create_using=nx.Graph())
print("Nodes:", G.number_of_nodes())
print("Edges:", G.number_of_edges())
G.nodes
```

Draw the graph

```python
plt.figure()
pos = nx.spring_layout(G, seed=0)
nx.draw(G, pos, with_labels=True, node_size=600)
plt.title("Message Network")
plt.show()
```

Degree and degree centrality

```python
deg = dict(G.degree())
deg_c = nx.degree_centrality(G)
deg_df = pd.DataFrame({
    'node': list(deg.keys()),
    'degree': list(deg.values()),
    'degree_centrality': [deg_c[n] for n in deg.keys()]
}).sort_values('degree', ascending=False)
deg_df
```

Betweenness and clustering

```python
btw = nx.betweenness_centrality(G)
clust = nx.clustering(G)
cent = pd.DataFrame({
    'node': list(G.nodes()),
    'betweenness': [btw[n] for n in G.nodes()],
    'clustering': [clust[n] for n in G.nodes()]
}).sort_values('betweenness', ascending=False)
cent
```

Subgraphs: ego network around a node

```python
ego = nx.ego_graph(G, 'I', radius=1)
pos_ego = nx.spring_layout(ego, seed=1)
plt.figure()
nx.draw(ego, pos_ego, with_labels=True, node_size=600)
plt.title("Ego network of node 'C'")
plt.show()
```

Connected components

```python
components = list(nx.connected_components(G))
comp_sizes = [len(c) for c in components]
pd.DataFrame({'component_id': range(1, len(components)+1), 'size': comp_sizes})
```

### Supervised Learning: Quick Start

First create some fake data after we bring in a lot of packages that we need.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

np.random.seed(123)
n = 400
X1 = np.random.normal(size=n)
X2 = np.random.normal(size=n)
# True probability uses a logistic function of X1 and X2
p = 1/(1+np.exp(-(0.5 + 1.2*X1 - 0.8*X2)))
y = (np.random.rand(n) < p).astype(int)

data = pd.DataFrame({'x1':X1, 'x2':X2, 'y':y})
data.head()
```

```python
data.describe()
```

Then we need to split our data into training and testing data. 30% goes into our testing data, a good standard approach.

```python
train, test = train_test_split(data, test_size=0.3, random_state=1)
```

```python
train
```

```python
test
```

Now we can model a logistic regression using 

```python
logit = LogisticRegression() # an empty logit model
logit.fit(train[['x1','x2']], train['y']) # pass the ivs of x1 and x2 and the dv of y to the model
pred_lr = logit.predict(test[['x1','x2']]) # try and predict the labels of x1 and x2 in the test model
```

How much did our model correctly predict?

```python
print("Logistic Regression Accuracy:", accuracy_score(test['y'], pred_lr))
```

A confusion matrix shows:
<br>True Negatives (TN): Predicted 0, actually 0 | False Positives (FP): Predicted 1, actually 0
<br>False Negatives (FN): Predicted 0, actually 1 | True Positives (TP): Predicted 1, actually 1

```python
print(confusion_matrix(test['y'], pred_lr))
```

```python
print(classification_report(test['y'], pred_lr))
```

Decision tree vs. logit

```python
tree = DecisionTreeClassifier(random_state=1234, max_depth=3)
tree.fit(train[['x1','x2']], train['y'])
pred_tree = tree.predict(test[['x1','x2']])
```

```python
print("Decision Tree Accuracy:", accuracy_score(test['y'], pred_tree))
```

```python
print(confusion_matrix(test['y'], pred_tree))
```

```python
print(classification_report(test['y'], pred_tree))
```

### Unsupervised Learning: k-means Clustering

```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans

np.random.seed(1234)
x = np.concatenate([np.random.normal(loc=-2, scale=1, size=100),
                    np.random.normal(loc= 3, scale=1, size=100)])
y = np.concatenate([np.random.normal(loc= 0, scale=1, size=100),
                    np.random.normal(loc= 0, scale=1, size=100)])
X = pd.DataFrame({'x':x,'y':y})
X.head()
```

```python
X.describe()
```

```python
inertias = []
ks = [2,3,4,5,6,7,8]
for k in ks:
    km = KMeans(n_clusters=k, n_init="auto", random_state=0)
    km.fit(X)
    inertias.append(km.inertia_)

inertia_table = pd.DataFrame({'k':ks, 'inertia':inertias})
inertia_table
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.plot(inertia_table['k'], inertia_table['inertia'], marker='o')
plt.title("Elbow plot (lower inertia is better)")
plt.xlabel("k (number of clusters)")
plt.ylabel("Inertia")
plt.show()
```

```python
km2 = KMeans(n_clusters=5, n_init=10, random_state=0)
labels = km2.fit_predict(X)
centers = km2.cluster_centers_

plt.figure()
plt.scatter(X['x'], X['y'], c=labels, alpha=0.7)
plt.scatter(centers[:,0], centers[:,1], marker='x', s=200) # adds an X to the "middle" of the cluster
plt.title("k-means with k=5")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```
